````python
# File: simplified_embed_chunk_store.py

import os
import json
import torch
from google.colab import drive
from langchain_community.vectorstores import Chroma
from langchain_core.embeddings import Embeddings
from langchain_core.documents import Document
from sklearn.metrics.pairwise import cosine_similarity


def mount_drive():
    drive.mount('/content/drive')

def load_json_files_from_folder(folder_path):
    all_chunks = []
    for filename in os.listdir(folder_path):
        if filename.endswith(".json"):
            file_path = os.path.join(folder_path, filename)
            with open(file_path, "r", encoding="utf-8") as file:
                try:
                    chunks = json.load(file)
                    all_chunks.extend([(chunk, filename) for chunk in chunks])
                except json.JSONDecodeError:
                    print(f"❌ Error reading {filename}")
    return all_chunks

class LlamaEmbeddings:
    def __init__(self, base_url='http://localhost:11434', model="nomic-embed-text"):
        from langchain_community.embeddings import OllamaEmbeddings
        #from langchain_ollama import OllamaEmbeddings
        self.model = OllamaEmbeddings(base_url=base_url, model=model)

    def embed_query(self, text: str) -> list:
        return self.model.embed_query(text)

    def embed_documents(self, texts: list) -> list:
        return self.model.embed_documents(texts)

def add_documents_to_chromadb(vectorstore, chunks_with_meta):
    try:
        documents = [
            Document(page_content=text, metadata={"source_file": meta})
            for text, meta in chunks_with_meta
        ]

        vectorstore.add_documents(documents)
        vectorstore.persist()
        print(f"✅ Added {len(documents)} documents to ChromaDB.")
        torch.cuda.empty_cache()
    except Exception as e:
        print(f"❌ Error: {e}")

def retrieve_all_documents(vectorstore, k=900):
    retriever = vectorstore.as_retriever()
    #docs = retriever.get_relevant_documents(query="", k=k)
    docs = retriever.invoke("")
    return docs

def main():
    global vectorstore, all_chunks_with_meta, embedding_model
    mount_drive()
    folder_path = "/content/drive/MyDrive/jason dimensions social"
    all_chunks_with_meta = load_json_files_from_folder(folder_path)

    embedding_model = LlamaEmbeddings(base_url="http://localhost:11434", model="nomic-embed-text")
    vectorstore = Chroma(persist_directory="./chroma_db", embedding_function=embedding_model)

    add_documents_to_chromadb(vectorstore, all_chunks_with_meta)

    stored_chunks = retrieve_all_documents(vectorstore)
    print("\n✅ Retrieved Chunks from ChromaDB:")
    for idx, doc in enumerate(stored_chunks, 1):
        print(f"Chunk {idx} ({doc.metadata}):\n{doc.page_content}\n{'-'*80}\n")

if __name__ == "__main__":
    main()
````

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# --------------------
# 🔹 Cosine similarity
# --------------------
def cosine_sim(a, b):
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-9))

def compute_cosine(a, b):
    return float(cosine_similarity([a], [b])[0][0])


# --------------------
# 🔹 RAG Function
# --------------------
def simple_rag_query_with_similarity(question, vectorstore, llm, embed_model, k=4):
    global answer, docs, cosine_scores, full_context
    try:
        retriever = vectorstore.as_retriever(search_kwargs={"k": k})
        docs = retriever.invoke(question)
        print("Retrieved docs:", len(docs))

        for i, d in enumerate(docs):
            print(f"Doc {i+1} length:", len(d.page_content))

        q_emb = np.array(embed_model.embed_query(question))
        cosine_scores = []
        context_parts = []

        for i, doc in enumerate(docs):
            chunk = doc.page_content
            c_emb = np.array(embed_model.embed_query(chunk))
            sim = cosine_sim(q_emb, c_emb)

            cosine_scores.append(sim)
            context_parts.append(f"[Document {i+1} | Cosine={sim:.4f}]: {chunk}")

        full_context = "\n\n".join(context_parts)

        prompt = f"""
You are an expert analyst. Based on the documents, answer the question thoroughly.

QUESTION: {question}

DOCUMENTS:
{full_context}

Please answer ONLY based on these documents.
"""

        answer = llm.invoke(prompt)

        return {
            "answer": answer,
            "docs": docs,
            "cosine_scores": cosine_scores,
            "context": full_context
        }

    except Exception as e:
        return {"error": str(e)}


# --------------------
# 🔹 Compute similarities
# --------------------
def compute_answer_similarities(answer, question, context, embedding_model):
    answer_vec = embedding_model.embed_query(answer)
    question_vec = embedding_model.embed_query(question)
    context_vec = embedding_model.embed_query(context)

    return {
        "answer_question_similarity": compute_cosine(answer_vec, question_vec),
        "answer_doc_similarity": compute_cosine(answer_vec, context_vec)
    }
````

````python
# collabeval_judge.py
# Clean, stable, weighted multi-agent judge with Ollama
import re
import time
import numpy as np
from typing import List, Dict, Any, Optional, Callable
from concurrent.futures import ThreadPoolExecutor, as_completed
from langchain_community.llms import Ollama

###############################################################################
#
#LLM-as-a-Judge
#Zheng et al., 2023 — “LLM as a Judge”
#Liu et al., 2023 — “G-EVAL”
#Chiang et al., 2024 — “AI Feedback for Evaluation” (OpenAI)
#Multi-Agent Judging / Debate
#Du et al., 2024 — “Multi-Agent Debate Improves LLM Judgments”
#OpenAI — “Model Spec & AI Debate Experiments”
#Multi-LLM Committee Studies (2023–2024)
###############################################################################

def safe_float(s: str, default=None):
    try:
        return float(s.strip())
    except:
        return default

###############################################################################
# Agent
###############################################################################

class Agent:
    def __init__(self, model_name: str, persona: str = "", num_ctx: int = 8192, temperature: float = 0):
        self.model_name = model_name
        self.persona = persona.strip()
        self.llm = Ollama(model=self.model_name, num_ctx=num_ctx, temperature=temperature, keep_alive="5m")

    def _compose(self, prompt: str) -> str:
        if self.persona:
            return f"{self.persona}\n\nUser: {prompt}"
        return prompt

    def ask(self, prompt: str) -> str:
        try:
            return self.llm.invoke(self._compose(prompt)).strip()
        except Exception:
            return ""

    def score_numeric(self, rubric_prompt: str) -> Optional[float]:
        out = self.ask(rubric_prompt)
        m = re.search(r"\d*\.?\d+", out)
        if not m:
            return None
        return max(0.0, min(1.0, float(m.group(0))))  # clamp 0–1


###############################################################################
# CollabEval Judge
###############################################################################

class CollabEvalJudge:
    """
    Multi-agent judge with weighted aggregation.
    """

    def __init__(
        self,
        agents_config: List[Dict[str, str]],
        rounds: int = 0,
        aggregation: str = "mean",  # "mean" or "weighted"
        weights: Optional[List[float]] = None,
        unanimity_threshold: float = 0.7,
        debate_prompt_template: Optional[Callable] = None
    ):
        self.agents = [Agent(cfg["model"], persona=cfg.get("persona","")) for cfg in agents_config]
        self.rounds = rounds
        self.aggregation = aggregation
        self.weights = np.array(weights) if weights is not None else None
        self.unanimity_threshold = float(unanimity_threshold)
        self.debate_prompt_template = debate_prompt_template or self.default_debate_prompt

    ###########################################################################
    # Prompt builder
    ###########################################################################

    @staticmethod
    def build_rubric_prompt(question: str, candidate: str, context: str, rubric: str):
        return f"""
Question: {question}

Context:
{context}

Candidate Answer:
{candidate}

Rubric:
{rubric}

ONLY OUTPUT A SINGLE NUMBER BETWEEN 0 AND 1. NO EXPLANATION.
"""

    @staticmethod
    def default_debate_prompt(candidate: str, prev_scores: List[float], prev_explanations: List[str], agent_id: str):
        others = "\n\n".join(
            [f"Agent{i}: score={s:.3f}\nExplanation: {e}" for i,(s,e) in enumerate(zip(prev_scores, prev_explanations))]
        )
        return f"""
You are Agent {agent_id}.
Candidate answer:
{candidate}

Other agents scored:
{others}

Revise ONLY if logically justified. Output ONLY the numeric score.
"""

    ###########################################################################
    # Parallel executor
    ###########################################################################

    def parallel_call(self, fn, agents, *args):
        results = [None] * len(agents)
        with ThreadPoolExecutor() as ex:
            futures = {ex.submit(fn, ag, *args): idx for idx, ag in enumerate(agents)}
            for f in as_completed(futures):
                idx = futures[f]
                try:
                    results[idx] = f.result()
                except:
                    results[idx] = None
        return results

    ###########################################################################
    # Main evaluation
    ###########################################################################

    def evaluate(
        self,
        question: str,
        candidate: str,
        context: str,
        rubric: str,
        aggregation: str = "mean"
    ) -> Dict[str, Any]:

        n_agents = len(self.agents)
        per_agent = [{"id": ag.model_name, "score": None} for ag in self.agents]

        rubric_prompt = self.build_rubric_prompt(question, candidate, context, rubric)

        # get raw numeric scores
        raw_scores = self.parallel_call(lambda agent, p: agent.score_numeric(p), self.agents, rubric_prompt)
        print("Raw scores:", raw_scores)

        # clean scores (replace None by 0)
        for i, s in enumerate(raw_scores):
            per_agent[i]["score"] = float(max(0.0, min(1.0, s if s is not None else 0.0)))

        scores = np.array([pa["score"] for pa in per_agent])

        #######################################################################
        # Weighted mean logic (YOUR REQUIREMENT)
        #######################################################################
        if aggregation == "weighted" and self.weights is not None:
            w = np.array(self.weights, dtype=float)
            if len(w) != len(scores):
                raise ValueError(f"Number of weights ({len(w)}) does not match number of agents ({len(scores)}).")
            w = w / np.sum(w)  # normalize
            final_score = float(np.sum(scores * w))

        else:
            final_score = float(np.mean(scores))

        #######################################################################
        # Confidence
        #######################################################################
        agree_frac = float(np.sum(np.abs(scores - final_score) <= 0.2) / n_agents)
        high_confidence = agree_frac >= self.unanimity_threshold

        return {
            "per_agent": per_agent,
            "aggregated_score": final_score,
            "agreement_fraction": agree_frac,
            "high_confidence": high_confidence,
        }


###############################################################################
# Adapter (optional)
###############################################################################

class CollabEvalAdapter:
    def __init__(self, judge: CollabEvalJudge, detect_mode_fn: Optional[Callable[[str], str]] = None):
        self.judge = judge
        self.detect_mode_fn = detect_mode_fn or self.simple_detect_mode

    def simple_detect_mode(self, prompt: str) -> str:
        P = prompt.upper()
        if "RATE FROM 0 TO 1" in P or "SCORE" in P:
            return "score"
        return "text"

    def invoke(self, prompt: str) -> str:
        mode = self.detect_mode_fn(prompt)
        if mode == "score":
            res = self.judge.evaluate(
                question="Scoring request",
                candidate="Candidate output",
                context="",
                rubric=prompt,
                aggregation=self.judge.aggregation
            )
            return f"{res['aggregated_score']:.4f}"
        return "Adapter only supports scoring mode."
````

````python
import numpy as np
import nltk

# ---------------------------------------------------
# Clamp numeric score between 0 and 1
# ---------------------------------------------------
def clamp01(x):
    try:
        x = float(x)
        return max(0.0, min(1.0, x))
    except:
        return 0.0

# ======================================================
# context_relevance_dual (بازنویسی شده برای بازگرداندن میانگین ساده همه امتیازات خام)
# ======================================================
def context_relevance_dual(
    question, answer,
    chunks,
    llm,
    prompt_template,
    debug=False,
    debug_store=None,
    repeats=0, # فرض می‌کنیم repeats=1 است (اگرچه شما 0 گذاشته بودید، منطقا باید >0 باشد)
    threshold=0.7,
):
    """
    Version WITHOUT splitting chunks.
    این نسخه، میانگین ساده تمام امتیازات خام (8 عدد در سناریوی 4 چانک و 2 تکرار/مدل)
    را به عنوان خروجی اصلی برمی‌گرداند.
    """

    if prompt_template is None:
        raise ValueError("prompt_template is required (e.g., PROMPT_IS_RELEVANT).")

    precision_scores_to_average = []
    true_positives = 0

    # 🌟 NEW: متغیرهایی برای محاسبه میانگین ساده
    all_raw_scores_sum = 0
    all_raw_scores_count = 0
    raw_avg_scores_per_chunk = [] # برای نگهداری میانگین خام هر چانک

    if debug and debug_store is not None:
        debug_store["context_relevance_dual"] = {"per_chunk_robustness": [], "steps": []}

    for i, ch in enumerate(chunks):

        # 🔵 chunk کامل استفاده می‌شود
        full_chunk = getattr(ch, "page_content", str(ch))
        parts = [full_chunk]

        local_scores = []

        for half in parts:
            half_scores = []

            for _ in range(repeats):
                prompt = prompt_template.format(question=question, chunk=half)
                resp = llm.invoke(prompt)

                score = clamp01(resp)
                half_scores.append(score)

                # 🌟 جمع کردن امتیاز خام برای محاسبه میانگین ساده کل
                all_raw_scores_sum += score
                all_raw_scores_count += 1

            half_avg = sum(half_scores) / repeats
            local_scores.append(half_avg)

        # محاسبه میانگین خام چانک (avg_relevance)
        avg_relevance = sum(local_scores) / len(local_scores)
        raw_avg_scores_per_chunk.append(avg_relevance) # ذخیره میانگین خام چانک

        is_robustly_relevant = avg_relevance >= threshold

        # منطق Robust Rank-Aware Precision (برای دیباگ و محاسبات داخلی حفظ می‌شود)
        if is_robustly_relevant:
            true_positives += 1
            current_precision = true_positives / (i + 1)
            precision_scores_to_average.append(current_precision)
            precision_str = f"{current_precision:.4f}"
        else:
            precision_str = "N/A"

        if debug:
            step_info = {
                "chunk_index": i,
                "avg_relevance": avg_relevance,
                "robustly_relevant": is_robustly_relevant,
                "current_tp": true_positives,
                "precision": precision_str
            }
            if debug_store is not None:
                debug_store["context_relevance_dual"]["per_chunk_robustness"].append(avg_relevance)
                debug_store["context_relevance_dual"]["steps"].append(step_info)
            print(f"[context_relevance_dual] Chunk {i+1} | Avg={avg_relevance:.4f} | TP={true_positives} | Precision={precision_str}")

    # محاسبه Robust Rank-Aware Precision (final_score_RAP)
    final_score_RAP = sum(precision_scores_to_average) / len(precision_scores_to_average) if precision_scores_to_average else 0.0

    # 🌟 محاسبه میانگین ساده تمام امتیازات خام (خروجی درخواستی شما)
    if all_raw_scores_count > 0:
        simple_mean_of_all_raw_scores = all_raw_scores_sum / all_raw_scores_count
    else:
        simple_mean_of_all_raw_scores = 0.0

    if debug and debug_store is not None:
        debug_store["context_relevance_dual"]["final_RAP"] = final_score_RAP
        debug_store["context_relevance_dual"]["final_SimpleMean"] = simple_mean_of_all_raw_scores
        print(f"[context_relevance_dual] FINAL Robust Rank-Aware Precision: {final_score_RAP:.4f}")
        print(f"[context_relevance_dual] FINAL Simple Mean of All Raw Scores: {simple_mean_of_all_raw_scores:.4f}")

    # 🌟 خروجی اصلی: میانگین ساده تمام امتیازات خام
    # خروجی دوم: لیست میانگین خام هر چانک
    return simple_mean_of_all_raw_scores
````

````python
def response_relevancy_llm(
    question: str,
    answer: str,
    llm, # Changed to a regular parameter
    runs: int = 5,
    debug: bool = False,
    debug_store: dict = None,
    prompt_template: str  = None,
) :
    """
    Runs multi-agent relevance scoring using CollabEvalAdapter.
    Returns aggregated score (0-1) and prints in notebook.
    """
    scores = []
    for i in trange(runs, desc="response_relevancy_llm"):
      prompt = prompt_template.format(question=question, answer=answer)
      resp = llm.invoke(prompt)
      try:
            score = float(resp)
      except:
            score = 0.0

        # محدود به 0 تا 1
      score = max(0.0, min(1.0, score))
      scores.append(score)

    avg_score = sum(scores) / len(scores) if scores else 0.0

    if debug:
        print("=" * 50)
        print("Scores:", scores)
        print("Average Score:", avg_score)
        print("=" * 50)

    return avg_score




````

````python
def evaluate_alignment_consistency_ret_question(
        eval_llm,
        question,
        chunks,
        runs=3,
        debug=False,
        debug_store=None,
        grand_list=None
    ):

    if not chunks:
        context_text = ""
    else:
        context_text = "\n\n".join([
            getattr(c, "page_content", str(c)) for c in chunks
        ])

    scores = []

    if debug and debug_store is not None:
        debug_store["alignment_ret_question"] = {"scores": []}

    for _ in trange(runs, desc="alignment_ret_question", leave=False):

        prompt = PROMPT_ALIGNMENT_QUESTION_TEST.format(
            question=str(question),
            contexts=context_text
        )

        # ---- CALL LLM ----
        resp = eval_llm.invoke(prompt)
        print("RAW:", resp)

        # ---- SAFE PARSE ----
        val = safe_float(resp)
        if val is not None:
            scores.append(val)
        else:
            print("⚠️ Non-numeric output ignored:", resp)

        if debug and debug_store is not None:
            debug_store["alignment_ret_question"]["scores"].append(val)

    arr = np.array(scores, dtype=float)
    mean_val = float(arr.mean()) if len(arr) else None

    # ✅ فقط 2 خط اضافه‌شده
    if grand_list is not None and mean_val is not None:
        grand_list.append(mean_val)
        grand_mean = float(np.mean(grand_list))
    else:
        grand_mean = None

    return  mean_val
````

````python
"""
judge_batch_v2.py  -  batch LLM-as-judge (3 metrics, LOCAL Ollama judges) over a ZIP of CSV files.

What it does when you run it
  1. asks you for the ZIP (Colab upload dialog) - or uses ZIP_PATH if you set it
  2. extracts it, finds EVERY .csv inside (any depth), and maps the columns automatically:
        question : question | query | prompt            (first one found, in this order)
        answer   : answer | response
        docs     : top_k_docs | retrieved_docs | ...      (chunks separated by |||)
     a column-mapping table is printed and saved; a file whose columns cannot be mapped is reported and skipped
  3. judges every file with 3 metrics:   ctx_answer, ctx_question, answer_question
  4. every judge's own score is saved separately:   gemma__ctx_answer, llama__ctx_answer, ...   (+ the aggregate)
  5. saves everything in a Google Drive folder (mirrors the folder structure of the zip):
        OUT_DIR/<same sub-folders>/<file>__judged.csv     one result file per input file
        OUT_DIR/_manifest.csv            what happened to each file (column mapping, rows, failures, time)
        OUT_DIR/_column_mapping.csv      which original column was used as question / answer / docs
        OUT_DIR/_debug_failures.jsonl    every unparsable / failed judge reply (raw text kept)
        OUT_DIR/_ALL_wide.csv  _ALL_long.csv  _summary.csv      all files together
  The run is resumable: finished rows / files are skipped when you run it again.

Colab:
    !ollama pull gemma3 ; !ollama pull llama3.1        # names must match JUDGES below (see `ollama list`)
    %run -i /content/judge_batch_v2.py
Optional, define BEFORE %run -i :
    LIMIT_FILES = 1 ; LIMIT_ROWS = 2        # quick test on one file, two rows
    ZIP_PATH = "/content/drive/MyDrive/xxx.zip"
    CONTEXT_MODE = "per_chunk"              # default "all" (all chunks in one prompt)
"""
import os, re, io, json, time, zipfile
from pathlib import Path
import numpy as np
import pandas as pd
from concurrent.futures import ThreadPoolExecutor
from langchain_ollama import OllamaLLM
from tqdm.auto import tqdm

# ----------------------------------------------------------------------------- config
JUDGES = {"deepseek": "deepseek-r1", "llama": "llama3.2"}   # label (used in column names) -> LOCAL Ollama model name
WEIGHTS = None                                       # e.g. {"gemma": 0.6, "llama": 0.4}; None = equal
OLLAMA_URL = "http://localhost:11434"
NUM_CTX = 16384                                      # prompts reach ~10k tokens (4 chunks + long answer)
TEMPERATURE = 0.0
REPEATS = 1                                          # >1 only useful with TEMPERATURE > 0
PARALLEL = False                                     # True = ask both judges at once (needs GPU memory for both)
CONTEXT_MODE = globals().get("CONTEXT_MODE", "all")  # "all" | "per_chunk"
CHUNK_SEP = "|||"
EXPECTED_CHUNKS = 4

ZIP_PATH = globals().get("ZIP_PATH")                 # None -> upload dialog
EXTRACT_DIR = globals().get("EXTRACT_DIR", "/content/judge_input")
OUT_DIR = globals().get("OUT_DIR", "/content/drive/MyDrive/judge_results_3metrics")
LIMIT_FILES = globals().get("LIMIT_FILES")
LIMIT_ROWS = globals().get("LIMIT_ROWS")

# accepted column names (compared lower-case, spaces -> "_"); first match in each list wins
ALIASES = {
    "question": ["question", "query", "prompt", "questions"],
    "answer":   ["answer", "response", "answers", "generated_answer"],
    "docs":     ["top_k_docs", "retrieved_docs", "retrieved_context", "retrieved_contexts", "contexts", "context", "docs"],
}

METRICS = ("ctx_answer", "ctx_question", "answer_question")
META_COLS = ["source_file", "combo", "generator", "embedder", "domain", "variation"]

# ----------------------------------------------------------------------------- prompts (texts unchanged from your code)
PROMPT_CTX_QUESTION_ALL = """Role: You are a Context Relevance Expert evaluating retrieval quality.
Task: Assess how well the retrieved context supports answering the user's question.
Criteria: Judge the degree to which the information in the retrieved context is relevant, factual, and sufficient to produce a complete and accurate answer to the question.

Question: {question}
Retrieved Context: {contexts}

Output Format: Provide a score from 0.0 (Irrelevant/Useless) to 1.0 (Highly relevant and sufficient). Output only the number.
"""

PROMPT_CTX_ANSWER_ALL = """Role: You are a Grounding Analyst evaluating factual support.
Task: Assess how well the generated answer is grounded in the retrieved context.
Criteria: Judge the degree to which all claims in the answer are logically and factually supported by the entire retrieved context.

Retrieved Context: {contexts}
Generated Answer: {answer}

Output Format: Provide a score from 0.0 (Ungrounded/Hallucinated) to 1.0 (Fully supported). Output only the number.
"""

PROMPT_CTX_QUESTION_CHUNK = """Role: You are a Chunk Relevance Assessor.
Task: Evaluate the individual relevance of the chunk to the user's question.
Criteria: Judge how essential the information in the chunk is for producing a correct and complete answer to the question.

Question: {question}
Retrieved Chunk: {chunk}

Output Format: Provide a score from 0.0 (Irrelevant) to 1.0 (Highly essential). Output only the number.
"""

PROMPT_CTX_ANSWER_CHUNK = """Role: You are a Context Utility Validator.
Task: Determine whether the retrieved chunk contributed meaningfully to the generated answer.
Criteria: Score based on explicit factual or semantic overlap between the chunk and the answer.

Retrieved Chunk: {chunk}
Generated Answer: {answer}

Output Format: Provide a score from 0.0 (No overlap/Not used) to 1.0 (Directly relevant and reflected in the answer). Output only the number.
"""

PROMPT_ANSWER_QUESTION = """Role: You are an Answer Quality Judge evaluating relevance.
Task: Assess whether the generated answer directly and meaningfully addresses the question.
Criteria: A high score means the answer accurately, directly, and thoroughly responds to the core intent of the question.

Question: {question}
Generated Answer: {answer}

Output Format: Provide a score from 0.0 (Completely off-topic) to 1.0 (Perfectly addresses the question). Output only the number.
"""


# ----------------------------------------------------------------------------- parsing
_THINK = re.compile(r"<think>.*?</think>", re.S)
_NUM = re.compile(r"(?<![\d.])(\d+(?:\.\d+)?|\.\d+)")


def parse_score(text):
    """First number in the reply; must be in [0, 1]. Anything else (empty, '8/10', no number) -> None."""
    if not text:
        return None
    m = _NUM.search(_THINK.sub("", str(text)).strip())
    if not m:
        return None
    v = float(m.group(1))
    return v if 0.0 <= v <= 1.0 else None


def parse_chunks(cell):
    return [c.strip() for c in str(cell).split(CHUNK_SEP) if c.strip()]


# ----------------------------------------------------------------------------- judge
class Judge:
    def __init__(self, judges, weights=None, num_ctx=NUM_CTX, temperature=TEMPERATURE,
                 repeats=REPEATS, parallel=PARALLEL, base_url=OLLAMA_URL):
        assert len(set(judges.values())) == len(judges), "use different models for different judges"
        self.labels, self.model_of = list(judges), dict(judges)
        self.repeats, self.num_ctx, self.parallel = repeats, num_ctx, parallel
        w = np.array([1.0 if weights is None else float(weights[l]) for l in self.labels])
        self.w = w / w.sum()
        self.llms = {l: OllamaLLM(model=judges[l], base_url=base_url, num_ctx=num_ctx,
                                  temperature=temperature, keep_alive="30m") for l in self.labels}
        self.n_calls = {l: 0 for l in self.labels}
        self.n_failed = {l: 0 for l in self.labels}
        self.failures = []          # raw replies that could not be parsed (debug)
        self.current = {}           # file / row currently being judged (copied into failure records)
        self._warned_len = False

    def _ask(self, label, prompt, tag):
        raw, err = None, None
        for attempt in range(3):                              # retry only real errors (connection, model busy)
            try:
                raw, err = str(self.llms[label].invoke(prompt)), None
                break
            except Exception as e:
                err = repr(e)
                time.sleep(2 * (attempt + 1))
        score = parse_score(raw) if raw is not None else None
        if score is None:
            self.failures.append({**self.current, "tag": tag, "judge": label, "model": self.model_of[label],
                                  "prompt_chars": len(prompt), "raw_reply": (raw or "")[:300], "error": err})
        return score

    def score(self, prompt, tag=""):
        """-> (aggregate score or None, {judge label: score or None})"""
        if not self._warned_len and len(prompt) > 3.0 * self.num_ctx:
            print(f"  WARNING: prompt has {len(prompt)} chars, may exceed num_ctx={self.num_ctx}; Ollama truncates silently.")
            self._warned_len = True
        workers = len(self.labels) * self.repeats if self.parallel else 1
        with ThreadPoolExecutor(max_workers=workers) as ex:
            futs = {l: [ex.submit(self._ask, l, prompt, tag) for _ in range(self.repeats)] for l in self.labels}
            per = {}
            for l, fl in futs.items():
                vals = [f.result() for f in fl]
                self.n_calls[l] += len(vals)
                self.n_failed[l] += sum(v is None for v in vals)
                vals = [v for v in vals if v is not None]
                per[l] = float(np.mean(vals)) if vals else None
        ok = [(i, per[l]) for i, l in enumerate(self.labels) if per[l] is not None]
        if not ok:
            return None, per
        w = np.array([self.w[i] for i, _ in ok])
        s = np.array([v for _, v in ok])
        return float((w * s).sum() / w.sum()), per


# ----------------------------------------------------------------------------- the three metrics
def _mean_or_none(xs):
    xs = [x for x in xs if x is not None]
    return float(np.mean(xs)) if xs else None


def score_context(judge, kind, text, chunks, mode=CONTEXT_MODE):
    """kind = 'question' or 'answer' (what the chunks are compared with)."""
    assert kind in ("question", "answer") and mode in ("all", "per_chunk")
    tag = f"ctx_{kind}"
    if mode == "all":
        tmpl = PROMPT_CTX_QUESTION_ALL if kind == "question" else PROMPT_CTX_ANSWER_ALL
        score, per = judge.score(tmpl.format(contexts="\n\n".join(chunks), **{kind: text}), tag)
        return {"score": score, "per_judge": per}
    tmpl = PROMPT_CTX_QUESTION_CHUNK if kind == "question" else PROMPT_CTX_ANSWER_CHUNK
    res = [judge.score(tmpl.format(chunk=c, **{kind: text}), f"{tag}[chunk{i}]") for i, c in enumerate(chunks, 1)]
    return {"score": _mean_or_none([s for s, _ in res]),
            "per_judge": {l: _mean_or_none([p[l] for _, p in res]) for l in judge.labels},
            "per_chunk": [s for s, _ in res]}


def score_answer_question(judge, question, answer):
    score, per = judge.score(PROMPT_ANSWER_QUESTION.format(question=question, answer=answer), "answer_question")
    return {"score": score, "per_judge": per}


def evaluate_row(judge, question, answer, chunks, mode=CONTEXT_MODE):
    return {"ctx_answer": score_context(judge, "answer", answer, chunks, mode),
            "ctx_question": score_context(judge, "question", question, chunks, mode),
            "answer_question": score_answer_question(judge, question, answer)}


def flatten(res, judge):
    """aggregate columns first, then ALL columns of judge 1, ALL columns of judge 2, ..., then disagreement."""
    rec = {k: res[k]["score"] for k in METRICS}
    for l in judge.labels:
        for k in METRICS:
            rec[f"{l}__{k}"] = res[k]["per_judge"][l]
    for k in METRICS:
        v = [res[k]["per_judge"][l] for l in judge.labels if res[k]["per_judge"][l] is not None]
        rec[f"{k}__judge_gap"] = (max(v) - min(v)) if len(v) > 1 else None      # |max - min| between judges
    if "per_chunk" in res["ctx_answer"]:
        for k in ("ctx_answer", "ctx_question"):
            rec[f"{k}__per_chunk"] = ";".join("" if s is None else f"{s:.3f}" for s in res[k]["per_chunk"])
    return rec


# ----------------------------------------------------------------------------- input: zip, files, columns
class ColumnMismatch(Exception):
    pass


def ensure_drive():
    if OUT_DIR.startswith("/content/drive") and not os.path.isdir("/content/drive/MyDrive"):
        from google.colab import drive
        drive.mount("/content/drive")


def get_zip():
    if ZIP_PATH:
        assert os.path.exists(ZIP_PATH), f"ZIP_PATH not found: {ZIP_PATH}"
        return ZIP_PATH
    from google.colab import files
    print("Select the ZIP file with all CSVs ...")
    up = files.upload()
    zips = [n for n in up if n.lower().endswith(".zip")]
    assert zips, "no .zip was uploaded"
    return zips[0]


def extract_zip(zip_path, dest=EXTRACT_DIR):
    os.makedirs(dest, exist_ok=True)
    with zipfile.ZipFile(zip_path) as z:
        z.extractall(dest)
    csvs = sorted(str(p) for p in Path(dest).rglob("*.csv")
                  if "__MACOSX" not in p.parts and not p.name.startswith("._") and not p.name.startswith("."))
    assert csvs, "no CSV found inside the zip"
    root = os.path.commonpath([os.path.dirname(p) for p in csvs]) if len(csvs) > 1 else os.path.dirname(csvs[0])
    return [(p, os.path.relpath(p, root)) for p in csvs]


def map_columns(cols):
    norm = {re.sub(r"\s+", "_", str(c).strip().lower()): c for c in cols}
    found, notes = {}, []
    for std, cands in ALIASES.items():
        hits = [c for c in cands if c in norm]
        if not hits:
            raise ColumnMismatch(f"no column for '{std}' (accepted: {cands}); columns are {list(cols)}")
        found[std] = norm[hits[0]]
        if len(hits) > 1:
            notes.append(f"{std}: used '{norm[hits[0]]}', ignored {[norm[h] for h in hits[1:]]}")
    return found, "; ".join(notes)


def read_csv_strict(path):
    try:
        return pd.read_csv(path, encoding="utf-8-sig")                       # strict: never silently drops lines
    except Exception as e:
        print(f"  strict read failed ({type(e).__name__}); retrying with python engine, bad lines reported")
        return pd.read_csv(path, encoding="utf-8-sig", engine="python", on_bad_lines="warn")


def load_standardized(path):
    raw = read_csv_strict(path)
    raw.columns = [str(c).strip() for c in raw.columns]
    mp, note = map_columns(raw.columns)
    df = pd.DataFrame({"row_id": range(len(raw)),
                       "question": raw[mp["question"]], "answer": raw[mp["answer"]], "docs": raw[mp["docs"]]})
    if "batch_id" in raw.columns:
        df["batch_id"] = raw["batch_id"]
    for c in ("question", "answer", "docs"):
        df[c] = df[c].where(df[c].notna(), "").astype(str).str.strip()
    n_in = len(df)
    df = df[(df.question != "") & (df.answer != "") & (df.docs != "")]
    return df, mp, note, n_in


def parse_meta(rel):
    parts = list(Path(rel).parts)
    folders = [p.lower() for p in parts[:-1]]
    allp = [p.lower() for p in parts]

    def pick(words, pool):
        for w in words:
            if any(w in p for p in pool):
                return w
        return ""
    combo = re.sub(r"\s+", "", parts[0]).lower() if len(parts) > 1 else ""
    domain = (pick(["ecolog"], folders) or pick(["ecolog"], allp)) and "ecological" or \
             ((pick(["social"], folders) or pick(["social"], allp)) and "social" or "")
    variation = (pick(["context"], folders) and "context") or (pick(["prompt"], folders) and "prompt") or \
                (pick(["context"], allp) and "context") or (pick(["prompt"], allp) and "prompt") or ""
    gen, _, emb = combo.partition("-")
    return {"source_file": rel, "combo": combo, "generator": gen, "embedder": emb,
            "domain": domain, "variation": variation}


# ----------------------------------------------------------------------------- run one file / all files
def out_path_for(rel):
    p = Path(rel)
    return os.path.join(OUT_DIR, str(p.parent), p.stem + "__judged.csv")


def process_file(judge, path, rel, mode=CONTEXT_MODE, limit_rows=None):
    meta = parse_meta(rel)
    df, mp, note, n_in = load_standardized(path)
    out_path = out_path_for(rel)
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    if limit_rows:
        df = df.head(limit_rows)

    rows, done = [], set()
    if os.path.exists(out_path):
        old = pd.read_csv(out_path)
        assert "row_id" in old.columns, f"{out_path} was not written by this script"
        rows, done = old.to_dict("records"), set(old["row_id"])
    todo = df[~df.row_id.isin(done)]
    t_file = time.time()
    n_fail0 = sum(judge.n_failed.values())
    for _, r in tqdm(todo.iterrows(), total=len(todo), desc=Path(rel).stem[:40], leave=False):
        chunks = parse_chunks(r["docs"])
        judge.current = {"source_file": rel, "row_id": int(r["row_id"])}
        t0 = time.time()
        res = evaluate_row(judge, r["question"], r["answer"], chunks, mode)
        rec = {**meta, "row_id": int(r["row_id"])}
        if "batch_id" in df.columns:
            rec["batch_id"] = r["batch_id"]
        rec.update({"n_chunks": len(chunks), "context_mode": mode, "seconds": round(time.time() - t0, 1),
                    **flatten(res, judge), "question": r["question"], "answer": r["answer"]})
        rows.append(rec)
        pd.DataFrame(rows).to_csv(out_path, index=False)          # saved after every row

    if judge.failures:
        with open(os.path.join(OUT_DIR, "_debug_failures.jsonl"), "a", encoding="utf-8") as f:
            for x in judge.failures:
                f.write(json.dumps(x, ensure_ascii=False) + "\n")
        judge.failures = []
    out = pd.DataFrame(rows)
    return {**meta,
            "col_question": mp["question"], "col_answer": mp["answer"], "col_docs": mp["docs"], "mapping_note": note,
            "rows_in_csv": n_in, "rows_valid": len(df), "rows_judged": len(out), "rows_new_this_run": len(todo),
            "chunks_min": int(out.n_chunks.min()) if len(out) else None,
            "chunks_max": int(out.n_chunks.max()) if len(out) else None,
            "failed_calls": sum(judge.n_failed.values()) - n_fail0, "seconds": round(time.time() - t_file, 1),
            "status": "ok", "output": out_path}


def build_outputs(judge):
    files = sorted(Path(OUT_DIR).rglob("*__judged.csv"))
    if not files:
        return
    wide = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
    wide.to_csv(os.path.join(OUT_DIR, "_ALL_wide.csv"), index=False)

    ids = META_COLS + ["row_id"]
    parts = []
    for k in METRICS:
        for j in ["aggregate"] + judge.labels:
            col = k if j == "aggregate" else f"{j}__{k}"
            if col in wide.columns:
                s = wide[ids + [col]].rename(columns={col: "score"})
                s["metric"], s["judge"] = k, j
                parts.append(s)
    pd.concat(parts, ignore_index=True).to_csv(os.path.join(OUT_DIR, "_ALL_long.csv"), index=False)

    score_cols = [c for c in wide.columns if c in METRICS or any(c == f"{l}__{k}" for l in judge.labels for k in METRICS)]
    summ = wide.groupby(META_COLS)[score_cols].agg("mean").round(4)
    summ.insert(0, "n_rows", wide.groupby(META_COLS).size())
    summ.reset_index().to_csv(os.path.join(OUT_DIR, "_summary.csv"), index=False)
    print(f"all-files outputs written ({len(files)} files, {len(wide)} rows) -> {OUT_DIR}")


def smoke_test(judge):
    """Tiny example: checks every judge is reachable and returns a parsable number."""
    q = "How does consumer demand contribute to modern slavery?"
    a = "Consumers create demand for goods produced through exploitation, even unknowingly, through price-driven choices."
    chunks = ["Consumers spend money on goods produced with the input of persons trapped in exploitation, "
              "thus knowingly or unknowingly feeding exploitation through their spending.",
              "Complex global supply chains mean that modern slavery can be less visible to the public eye."]
    res = evaluate_row(judge, q, a, chunks, "all")
    ok = True
    for k, v in res.items():
        print(f"  {k:16s} aggregate={v['score']}  per judge={ {l: (None if s is None else round(s, 3)) for l, s in v['per_judge'].items()} }")
    for l in judge.labels:
        if all(res[k]["per_judge"][l] is None for k in METRICS):
            ok = False
            print(f"  !! judge '{l}' ({judge.model_of[l]}) returned no valid number: is the model pulled? (`ollama list`)")
    judge.failures = []
    return ok


def run_all(judge):
    ensure_drive()
    os.makedirs(OUT_DIR, exist_ok=True)
    zip_path = get_zip()
    files = extract_zip(zip_path)
    if LIMIT_FILES:
        files = files[:LIMIT_FILES]
    print(f"\n{len(files)} CSV files found in the zip.\n")

    # 1) column check for ALL files before spending any GPU time
    plan, bad = [], []
    for path, rel in files:
        try:
            df, mp, note, n_in = load_standardized(path)
            plan.append({**parse_meta(rel), "col_question": mp["question"], "col_answer": mp["answer"],
                         "col_docs": mp["docs"], "rows_in_csv": n_in, "rows_valid": len(df), "note": note})
        except ColumnMismatch as e:
            bad.append({"source_file": rel, "status": "SKIPPED (columns)", "error": str(e)})
    plan_df = pd.DataFrame(plan)
    plan_df.to_csv(os.path.join(OUT_DIR, "_column_mapping.csv"), index=False)
    with pd.option_context("display.width", 250, "display.max_colwidth", 60, "display.max_rows", None):
        print(plan_df[["source_file", "col_question", "col_answer", "col_docs", "rows_in_csv", "rows_valid"]].to_string(index=False))
    print("\ncolumn names used:", {c: plan_df[c].value_counts().to_dict() for c in ("col_question", "col_answer", "col_docs")})
    for b in bad:
        print("  !!", b["source_file"], "->", b["error"])
    print(f"total rows to judge: {int(plan_df.rows_valid.sum())}  "
          f"(~{int(plan_df.rows_valid.sum()) * 3 * len(judge.labels)} judge calls in 'all' mode)\n")

    # 2) judges reachable?
    print("smoke test:")
    assert smoke_test(judge), "fix the judge models first"

    # 3) judge every file
    manifest = list(bad)
    for path, rel in tqdm(files, desc="files"):
        if any(b["source_file"] == rel for b in bad):
            continue
        try:
            manifest.append(process_file(judge, path, rel, CONTEXT_MODE, LIMIT_ROWS))
        except Exception as e:
            print(f"  !! {rel}: {type(e).__name__}: {e}")
            manifest.append({"source_file": rel, "status": f"ERROR {type(e).__name__}", "error": str(e)})
        pd.DataFrame(manifest).to_csv(os.path.join(OUT_DIR, "_manifest.csv"), index=False)

    build_outputs(judge)
    man = pd.DataFrame(manifest)
    print("\n=== DONE ===")
    print(man["status"].value_counts().to_dict())
    print("failed/unparsable calls per judge:", {l: f"{judge.n_failed[l]}/{judge.n_calls[l]}" for l in judge.labels})
    print("results folder:", OUT_DIR)
    return man


# ----------------------------------------------------------------------------- main
if globals().get("RUN_ALL", True):
    JUDGE = Judge(JUDGES, WEIGHTS)
    MANIFEST = run_all(JUDGE)
````

````python
"""
rank_k_pipeline.py  -  ONE self-contained, clean pipeline (replaces rank_k_analysis.py + all the patch cells)

What it does
  1. Builds (or loads, if they already exist) three Chroma stores with ONE embedder and identical settings:
        eco_v3 , soc_v3 , mixed_v3        (chunk domain is stored in the metadata as "eco::<file>" / "soc::<file>",
                                            so no folder/filename matching is ever needed)
  2. Sanity checks (stored vectors == re-embedded full text, counts, domain tags)
  3. Experiment A  separate corpora : cosine of top-K per domain, delta = eco - soc, with an OFF-TOPIC CONTROL
                                      and the control-corrected delta
  4. Experiment B  mixed corpus     : eco share / rank-weighted exposure / first hit / domain at each rank vs K
  5. Saves CSV + PNG into OUT_DIR

Run in Colab (Ollama already running, `final_prompts` already defined = your 20 prompts):
        %run -i /content/rank_k_pipeline.py
"""
import os, json, time, random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from langchain_community.vectorstores import Chroma
from langchain_ollama import OllamaEmbeddings

# ----------------------------------------------------------------------------- config
SRC = {"eco": "/content/drive/MyDrive/jason dimensions ecological",
       "soc": "/content/drive/MyDrive/jason dimensions social"}          # folder with the 192 social JSONs
STORE_DIR = "/content/drive/MyDrive/vectorstore_v3"                      # stores persist here (reused on re-run)
OUT_DIR = "/content/drive/MyDrive/rank_k_results_v3"
K_LIST = [1, 2, 3, 4, 6, 8, 10, 15, 20]
K_MAX, BASE_K = max(K_LIST), 4
N_BOOT, SEED = 5000, 2026
BATCH = 32

os.makedirs(STORE_DIR, exist_ok=True)
os.makedirs(OUT_DIR, exist_ok=True)
rng = np.random.default_rng(SEED)

PROMPTS = globals().get("final_prompts")
assert PROMPTS and len(PROMPTS) >= 2, "define final_prompts (your 20 prompts) before running"
print("prompts:", len(PROMPTS))

# ONE embedder for documents AND queries (num_ctx so Ollama does not cut ~880-token chunks)
embedder = OllamaEmbeddings(model="nomic-embed-text", base_url="http://localhost:11434", num_ctx=8192)

NEUTRAL = [   # off-topic controls: any eco-vs-soc gap here is a store/corpus artifact, not a domain effect
    "football match results in the premier league", "recipe for chocolate cake with butter and sugar",
    "history of the Roman empire and its emperors", "how to train a neural network for image classification",
    "symptoms and treatment of the common cold", "rules of chess opening strategy",
    "plot summary of a classic detective novel", "how do airplanes stay in the air",
    "tips for learning a foreign language quickly", "history of the printing press",
    "how to change a flat tire on a car", "the life of a famous classical composer",
    "basic principles of personal budgeting", "how does a refrigerator work",
    "olympic swimming world records", "introduction to quantum mechanics",
    "how to brew good coffee at home", "architecture of gothic cathedrals",
    "how to write a cover letter for a job", "the history of the internet",
]


# ----------------------------------------------------------------------------- 1. build / load stores
def load_chunks(folder, dom):
    items = []
    for fn in sorted(f for f in os.listdir(folder) if f.lower().endswith(".json")):
        with open(os.path.join(folder, fn), encoding="utf-8") as f:
            for c in json.load(f):
                if isinstance(c, str) and len(c.strip()) > 50:            # same filter for both domains
                    items.append((c, f"{dom}::{fn}"))
    return items


def embed_all(texts):
    vecs = []
    for i in range(0, len(texts), BATCH):
        for attempt in range(4):
            try:
                vecs += embedder.embed_documents(texts[i:i + BATCH])
                break
            except Exception as e:
                print("  retry", attempt, e)
                time.sleep(2 + attempt)
        else:
            raise RuntimeError(f"embedding failed at batch starting {i}")
        if (i // BATCH) % 10 == 0:
            print(f"  embedded {len(vecs)}/{len(texts)}")
    return vecs


def open_store(name):
    return Chroma(persist_directory=f"{STORE_DIR}/{name}", collection_name=name,
                  embedding_function=embedder, collection_metadata={"hnsw:space": "cosine"})


def domain_store(dom):
    vs = open_store(f"{dom}_v3")
    if vs._collection.count() > 0:
        print(f"{dom}: loaded existing store ({vs._collection.count()} chunks)")
        return vs
    items = load_chunks(SRC[dom], dom)
    print(f"{dom}: embedding {len(items)} chunks from {len({s for _, s in items})} files ...")
    vecs = embed_all([t for t, _ in items])
    for i in range(0, len(items), 500):
        sl = slice(i, i + 500)
        vs._collection.add(ids=[f"{dom}_{j}" for j in range(i, min(i + 500, len(items)))],
                           documents=[t for t, _ in items[sl]],
                           metadatas=[{"source_file": s} for _, s in items[sl]],
                           embeddings=vecs[sl])
    return vs


def mixed_store(stores):
    vs = open_store("mixed_v3")
    if vs._collection.count() > 0:
        print(f"mixed: loaded existing store ({vs._collection.count()} chunks)")
        return vs
    for dom, s in stores.items():                                           # reuse vectors, no re-embedding
        d = s._collection.get(include=["documents", "metadatas", "embeddings"])
        vs._collection.add(ids=[f"mixed_{dom}_{i}" for i in d["ids"]], documents=d["documents"],
                           metadatas=d["metadatas"], embeddings=np.asarray(d["embeddings"]).tolist())
    return vs


eco_vs, soc_vs = domain_store("eco"), domain_store("soc")
mixed_vs = mixed_store({"eco": eco_vs, "soc": soc_vs})


# ----------------------------------------------------------------------------- 2. sanity checks
def cosine(a, b):
    a, b = np.asarray(a, float).reshape(-1), np.asarray(b, float).reshape(-1)
    d = np.linalg.norm(a) * np.linalg.norm(b)
    return float(a @ b / d) if d else 0.0


def check_store(vs, n=20):
    d = vs._collection.get(limit=300, include=["documents", "embeddings"])
    idx = random.sample(range(len(d["ids"])), min(n, len(d["ids"])))
    s = [cosine(embedder.embed_query(d["documents"][i]), d["embeddings"][i]) for i in idx]
    return float(np.mean(s))


def dom_of(meta):
    return str((meta or {}).get("source_file", "")).split("::")[0]


def domain_counts(vs):
    ms = vs._collection.get(include=["metadatas"])["metadatas"]
    doms = pd.Series([dom_of(m) for m in ms]).value_counts().to_dict()
    files = len({(m or {}).get("source_file") for m in ms})
    return doms, files


print("\n=== SANITY CHECKS ===")
for nm, vs in (("eco", eco_vs), ("soc", soc_vs), ("mixed", mixed_vs)):
    print(f"{nm:6s} chunks={vs._collection.count():5d}  domains/files={domain_counts(vs)}  "
          f"re-embed vs stored cosine={check_store(vs):.4f}   (expect ~1.0000)")
exp_counts = domain_counts(mixed_vs)[0]
EXPECTED_SHARE = exp_counts.get("eco", 0) / max(1, exp_counts.get("eco", 0) + exp_counts.get("soc", 0))
assert set(exp_counts) <= {"eco", "soc"}, f"unknown domain tags in mixed store: {exp_counts}"
print(f"expected eco share of the mixed index = {EXPECTED_SHARE:.3f}")


# ----------------------------------------------------------------------------- 3. retrieval (once per query)
def retrieve(vs, query, k=K_MAX):
    q = embedder.embed_query(query)
    r = vs._collection.query(query_embeddings=[q], n_results=k, include=["metadatas", "embeddings"])
    embs = np.asarray(r["embeddings"][0])
    return [{"rank": i, "dom": dom_of(m), "cos": cosine(q, e)}
            for i, (m, e) in enumerate(zip(r["metadatas"][0], embs), 1)]


def boot_ci(x, n=N_BOOT):
    x = np.asarray(pd.Series(x).dropna(), float)
    if len(x) < 2:
        return (float(x.mean()) if len(x) else np.nan), np.nan, np.nan
    m = rng.choice(x, (n, len(x))).mean(1)
    return float(x.mean()), float(np.quantile(m, .025)), float(np.quantile(m, .975))


# ----------------------------------------------------------------------------- 4. Experiment A: separate corpora
def run_separate(queries):
    rows = []
    for dom, vs in (("eco", eco_vs), ("soc", soc_vs)):
        for pid, q in enumerate(queries):
            hits = retrieve(vs, q)
            for K in K_LIST:
                c = [h["cos"] for h in hits[:K]]
                rows.append({"domain": dom, "prompt_id": pid, "K": K, "cos_mean": np.mean(c), "cos_rank1": hits[0]["cos"]})
    return pd.DataFrame(rows)


def paired_diffs(df):
    out = {}
    for K, g in df.groupby("K"):
        e = g[g.domain == "eco"].set_index("prompt_id")["cos_mean"]
        s = g[g.domain == "soc"].set_index("prompt_id")["cos_mean"]
        out[K] = (e - s).dropna().to_numpy()
    return out


def summarize_separate(df_task, df_ctrl):
    dt, dc = paired_diffs(df_task), paired_diffs(df_ctrl)
    rows = []
    for K in K_LIST:
        m, lo, hi = boot_ci(dt[K])
        mc, loc, hic = boot_ci(dc[K])
        corr = rng.choice(dt[K], (N_BOOT, len(dt[K]))).mean(1) - rng.choice(dc[K], (N_BOOT, len(dc[K]))).mean(1)
        rows.append({"K": K, "delta_task": m, "task_lo": lo, "task_hi": hi,
                     "delta_neutral_control": mc, "ctrl_lo": loc, "ctrl_hi": hic,
                     "delta_corrected": float(corr.mean()), "corr_lo": float(np.quantile(corr, .025)),
                     "corr_hi": float(np.quantile(corr, .975)), "P(corrected>0)": float((corr > 0).mean())})
    return pd.DataFrame(rows)


# ----------------------------------------------------------------------------- 5. Experiment B: mixed corpus
def w_dcg(n):
    return 1.0 / np.log2(np.arange(2, n + 2))


def run_mixed(queries):
    rows, rank_rows = [], []
    for pid, q in enumerate(queries):
        hits = retrieve(mixed_vs, q)
        dom = np.array([h["dom"] for h in hits])
        cos = np.array([h["cos"] for h in hits])
        for r, d in enumerate(dom, 1):
            rank_rows.append({"prompt_id": pid, "rank": r, "p_eco": d == "eco", "p_soc": d == "soc"})
        for K in K_LIST:
            d, c, w = dom[:K], cos[:K], w_dcg(K)
            e, s = d == "eco", d == "soc"
            rows.append({"prompt_id": pid, "K": K, "n_eco": int(e.sum()), "n_soc": int(s.sum()),
                         "eco_share": e.mean(), "eco_exposure": float((w * e).sum() / w.sum()),
                         "first_eco": (np.argmax(e) + 1) if e.any() else np.nan,
                         "first_soc": (np.argmax(s) + 1) if s.any() else np.nan,
                         "cos_eco": c[e].mean() if e.any() else np.nan,
                         "cos_soc": c[s].mean() if s.any() else np.nan})
    return pd.DataFrame(rows), pd.DataFrame(rank_rows)


def summarize_mixed(df):
    rows = []
    for K, g in df.groupby("K"):
        r = {"K": K, "expected_share": EXPECTED_SHARE}
        for c in ("eco_share", "eco_exposure", "first_eco", "first_soc"):
            r[c], r[c + "_lo"], r[c + "_hi"] = boot_ci(g[c])
        r["eco_share_excess"] = r["eco_share"] - EXPECTED_SHARE
        r["delta_cos_within_topK"] = float((g.cos_eco - g.cos_soc).dropna().mean())
        b = rng.choice(g.eco_share.to_numpy(), (N_BOOT, len(g))).mean(1)
        r["P(eco_share>expected)"] = float((b > EXPECTED_SHARE).mean())
        rows.append(r)
    return pd.DataFrame(rows)


# ----------------------------------------------------------------------------- run
print("\n=== EXPERIMENT A: separate corpora (task prompts + off-topic control) ===")
df_task = run_separate(PROMPTS)
df_ctrl = run_separate(NEUTRAL)
sum_sep = summarize_separate(df_task, df_ctrl)
print(sum_sep.round(4).to_string(index=False))

print("\n=== EXPERIMENT B: mixed corpus ===")
df_mix, df_rank = run_mixed(PROMPTS)
sum_mix = summarize_mixed(df_mix)
rank_sum = df_rank.groupby("rank")[["p_eco", "p_soc"]].mean().reset_index()
print(sum_mix.round(4).to_string(index=False))
print("\nP(chunk at rank r is ecological / social):")
print(rank_sum.round(3).to_string(index=False))

# save
df_task.to_csv(f"{OUT_DIR}/separate_task_raw.csv", index=False)
df_ctrl.to_csv(f"{OUT_DIR}/separate_control_raw.csv", index=False)
sum_sep.to_csv(f"{OUT_DIR}/separate_summary.csv", index=False)
df_mix.to_csv(f"{OUT_DIR}/mixed_raw.csv", index=False)
sum_mix.to_csv(f"{OUT_DIR}/mixed_summary.csv", index=False)
rank_sum.to_csv(f"{OUT_DIR}/rank_positions.csv", index=False)

# plots
fig, ax = plt.subplots(1, 3, figsize=(16, 4))
ax[0].errorbar(sum_sep.K, sum_sep.delta_task, yerr=[sum_sep.delta_task - sum_sep.task_lo, sum_sep.task_hi - sum_sep.delta_task],
               marker="o", capsize=3, label="task prompts")
ax[0].errorbar(sum_sep.K, sum_sep.delta_neutral_control, yerr=[sum_sep.delta_neutral_control - sum_sep.ctrl_lo,
               sum_sep.ctrl_hi - sum_sep.delta_neutral_control], marker="s", capsize=3, label="off-topic control")
ax[0].plot(sum_sep.K, sum_sep.delta_corrected, "k--", marker="^", label="corrected (task - control)")
ax[0].axhline(0, color="r", lw=.8); ax[0].axvline(BASE_K, color="gray", ls=":")
ax[0].set_xlabel("K"); ax[0].set_ylabel("Δ cosine (eco − soc)"); ax[0].set_title("Separate corpora"); ax[0].legend(fontsize=7)

ax[1].plot(sum_mix.K, sum_mix.eco_share, marker="o", label="eco share")
ax[1].fill_between(sum_mix.K, sum_mix.eco_share_lo, sum_mix.eco_share_hi, alpha=.2)
ax[1].plot(sum_mix.K, sum_mix.eco_exposure, marker="s", label="eco exposure (rank-weighted)")
ax[1].axhline(EXPECTED_SHARE, color="k", ls="--", label="index share (neutral)")
ax[1].axvline(BASE_K, color="gray", ls=":"); ax[1].set_xlabel("K"); ax[1].set_title("Mixed corpus: who wins top-K")
ax[1].legend(fontsize=7)

ax[2].bar(rank_sum["rank"], rank_sum.p_eco, label="eco"); ax[2].bar(rank_sum["rank"], rank_sum.p_soc, bottom=rank_sum.p_eco, label="social")
ax[2].set_xlabel("rank position"); ax[2].set_ylabel("share of prompts"); ax[2].set_title("Domain at each rank"); ax[2].legend(fontsize=7)
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/rank_k_figures.png", dpi=200); plt.show()
print("\nSaved to:", OUT_DIR)
````

````python
"""
rank_k_pipeline_v4.py  -  same experiment as rank_k_pipeline.py, but parameterised by embedding model + prefix mode,
                          with a cross-run comparison at the end.

Runs (default, edit RUNS to change):
    ("nomic-embed-text", False)   legacy run, no prefix        -> reuses your existing vectorstore_v3 / rank_k_results_v3
    ("nomic-embed-text", True)    nomic with search_query / search_document prefixes
    ("embeddinggemma",   True)    gemma with its official task prefixes
    ("embeddinggemma",   False)   gemma without prefixes (isolates the effect of the prefix)

Each run gets its own folders:   vectorstore_<tag>/   rank_k_results_<tag>/    (stores are NEVER shared between models)
A run whose results CSVs already exist is skipped and loaded from disk (set FORCE_RERUN = True before %run to recompute).

Colab usage (Ollama running; `final_prompts` already defined = your 20 prompts):
    !ollama pull embeddinggemma          # once; needs Ollama >= 0.11.10
    %run -i /content/rank_k_pipeline_v4.py

Optional overrides (define BEFORE %run -i):
    RUNS = [("embeddinggemma", True)]                       # run only some configs
    TAG_OVERRIDE = {("embeddinggemma", True): "embeddinggemma"}   # point a config at folders you already filled
    FORCE_RERUN = True
"""
import os, json, time, random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from langchain_core.embeddings import Embeddings
from langchain_community.vectorstores import Chroma
from langchain_ollama import OllamaEmbeddings

# ----------------------------------------------------------------------------- config
BASE = "/content/drive/MyDrive"
SRC = {"eco": f"{BASE}/jason dimensions ecological",
       "soc": f"{BASE}/jason dimensions social"}
OLLAMA_URL = "http://localhost:11434"
K_LIST = [1, 2, 3, 4, 6, 8, 10, 15, 20]
K_MAX, BASE_K = max(K_LIST), 4
N_BOOT, SEED = 5000, 2026
BATCH = 32

# per-model settings. num_ctx: nomic accepts 8192, embeddinggemma's maximum is 2048.
PRESETS = {
    "nomic-embed-text": dict(q="search_query: ", d="search_document: ", num_ctx=8192),
    "embeddinggemma":   dict(q="task: search result | query: ", d="title: none | text: ", num_ctx=2048),
}

RUNS = globals().get("RUNS") or [
    ("nomic-embed-text", False),
    ("nomic-embed-text", True),
    ("embeddinggemma", True),
    ("embeddinggemma", False),
]
TAG_OVERRIDE = {("nomic-embed-text", False): "v3",           # legacy run keeps its old folders
                **globals().get("TAG_OVERRIDE", {})}         # user-supplied overrides (defined before %run -i)
FORCE_RERUN = bool(globals().get("FORCE_RERUN", False))

PROMPTS = globals().get("final_prompts")
assert PROMPTS and len(PROMPTS) >= 2, "define final_prompts (your 20 prompts) before running"
print("prompts:", len(PROMPTS), "| runs:", RUNS)

NEUTRAL = [   # off-topic controls: any eco-vs-soc gap here is a store/corpus artifact, not a domain effect
    "football match results in the premier league", "recipe for chocolate cake with butter and sugar",
    "history of the Roman empire and its emperors", "how to train a neural network for image classification",
    "symptoms and treatment of the common cold", "rules of chess opening strategy",
    "plot summary of a classic detective novel", "how do airplanes stay in the air",
    "tips for learning a foreign language quickly", "history of the printing press",
    "how to change a flat tire on a car", "the life of a famous classical composer",
    "basic principles of personal budgeting", "how does a refrigerator work",
    "olympic swimming world records", "introduction to quantum mechanics",
    "how to brew good coffee at home", "architecture of gothic cathedrals",
    "how to write a cover letter for a job", "the history of the internet",
]


# ----------------------------------------------------------------------------- small helpers
def cosine(a, b):
    a, b = np.asarray(a, float).reshape(-1), np.asarray(b, float).reshape(-1)
    d = np.linalg.norm(a) * np.linalg.norm(b)
    return float(a @ b / d) if d else 0.0


def dom_of(meta):
    return str((meta or {}).get("source_file", "")).split("::")[0]


def boot_ci(x, rng, n=N_BOOT):
    x = np.asarray(pd.Series(x).dropna(), float)
    if len(x) < 2:
        return (float(x.mean()) if len(x) else np.nan), np.nan, np.nan
    m = rng.choice(x, (n, len(x))).mean(1)
    return float(x.mean()), float(np.quantile(m, .025)), float(np.quantile(m, .975))


def w_dcg(n):
    return 1.0 / np.log2(np.arange(2, n + 2))


def run_tag(model, use_prefix):
    if (model, use_prefix) in TAG_OVERRIDE:
        return TAG_OVERRIDE[(model, use_prefix)]
    return f"{model.replace('-', '_').replace(':', '_')}_{'pref' if use_prefix else 'nopref'}"


def load_chunks(folder, dom):
    items = []
    for fn in sorted(f for f in os.listdir(folder) if f.lower().endswith(".json")):
        with open(os.path.join(folder, fn), encoding="utf-8") as f:
            for c in json.load(f):
                if isinstance(c, str) and len(c.strip()) > 50:            # same filter for both domains
                    items.append((c, f"{dom}::{fn}"))
    return items


class PrefixedEmbedder(Embeddings):
    """Adds the model's query / document prefix. Queries go through base.embed_documents on purpose:
    langchain_ollama's embed_query itself calls embed_documents, so overriding both on a subclass would prefix twice."""
    def __init__(self, base, q_prefix="", d_prefix=""):
        self.base, self.qp, self.dp = base, q_prefix, d_prefix

    def embed_documents(self, texts):
        return self.base.embed_documents([self.dp + t for t in texts])

    def embed_query(self, text):
        return self.base.embed_documents([self.qp + text])[0]


# ----------------------------------------------------------------------------- one full run = one embedder config
class Pipeline:
    def __init__(self, model, use_prefix):
        assert model in PRESETS, f"add {model} to PRESETS"
        self.model, self.use_prefix = model, use_prefix
        self.tag = run_tag(model, use_prefix)
        self.label = f"{model} ({'prefix' if use_prefix else 'no prefix'})"
        self.store_dir = f"{BASE}/vectorstore_{self.tag}"
        self.out_dir = f"{BASE}/rank_k_results_{self.tag}"
        os.makedirs(self.store_dir, exist_ok=True)
        os.makedirs(self.out_dir, exist_ok=True)
        self.rng = np.random.default_rng(SEED)                  # same seed per run -> comparable bootstrap

        p = PRESETS[model]
        self.ctx = p["num_ctx"]
        base = OllamaEmbeddings(model=model, base_url=OLLAMA_URL, num_ctx=p["num_ctx"])
        self.emb = PrefixedEmbedder(base, p["q"] if use_prefix else "", p["d"] if use_prefix else "")
        try:
            self.dim = len(self.emb.embed_query("test"))
        except Exception as e:
            raise RuntimeError(f"cannot embed with '{model}'. Did you run `ollama pull {model}`? ({e})")
        print(f"\n{'=' * 90}\nRUN  {self.label}   tag={self.tag}   dim={self.dim}\n{'=' * 90}")

    # ---- stores (collection names are fixed; the folder separates runs)
    def open_store(self, name):
        return Chroma(persist_directory=f"{self.store_dir}/{name}", collection_name=name,
                      embedding_function=self.emb, collection_metadata={"hnsw:space": "cosine"})

    def embed_all(self, texts):
        vecs = []
        for i in range(0, len(texts), BATCH):
            for attempt in range(4):
                try:
                    vecs += self.emb.embed_documents(texts[i:i + BATCH])
                    break
                except Exception as e:
                    print("  retry", attempt, e)
                    time.sleep(2 + attempt)
            else:
                raise RuntimeError(f"embedding failed at batch starting {i}")
            if (i // BATCH) % 10 == 0:
                print(f"  embedded {len(vecs)}/{len(texts)}")
        return vecs

    def domain_store(self, dom):
        vs = self.open_store(f"{dom}_v3")
        if vs._collection.count() > 0:
            print(f"{dom}: loaded existing store ({vs._collection.count()} chunks)")
            return vs
        items = load_chunks(SRC[dom], dom)
        mx = max(len(t) for t, _ in items)
        print(f"{dom}: embedding {len(items)} chunks from {len({s for _, s in items})} files "
              f"(longest chunk {mx} chars) ...")
        if mx > 3.5 * self.ctx:
            print(f"  WARNING: longest chunk may exceed the {self.ctx}-token context of {self.model}; Ollama can truncate silently.")
        vecs = self.embed_all([t for t, _ in items])
        for i in range(0, len(items), 500):
            sl = slice(i, i + 500)
            vs._collection.add(ids=[f"{dom}_{j}" for j in range(i, min(i + 500, len(items)))],
                               documents=[t for t, _ in items[sl]],
                               metadatas=[{"source_file": s} for _, s in items[sl]],
                               embeddings=vecs[sl])
        return vs

    def mixed_store(self, stores):
        vs = self.open_store("mixed_v3")
        if vs._collection.count() > 0:
            print(f"mixed: loaded existing store ({vs._collection.count()} chunks)")
            return vs
        for dom, s in stores.items():                          # reuse vectors, no re-embedding
            d = s._collection.get(include=["documents", "metadatas", "embeddings"])
            vs._collection.add(ids=[f"mixed_{dom}_{i}" for i in d["ids"]], documents=d["documents"],
                               metadatas=d["metadatas"], embeddings=np.asarray(d["embeddings"]).tolist())
        return vs

    # ---- sanity checks (document path: stored vector must equal re-embedded document text)
    def check_store(self, vs, n=20):
        d = vs._collection.get(limit=300, include=["documents", "embeddings"])
        idx = random.sample(range(len(d["ids"])), min(n, len(d["ids"])))
        s = [cosine(self.emb.embed_documents([d["documents"][i]])[0], d["embeddings"][i]) for i in idx]
        return float(np.mean(s))

    def domain_counts(self, vs):
        ms = vs._collection.get(include=["metadatas"])["metadatas"]
        doms = pd.Series([dom_of(m) for m in ms]).value_counts().to_dict()
        return doms, len({(m or {}).get("source_file") for m in ms})

    # ---- retrieval
    def retrieve(self, vs, query, k=K_MAX):
        q = self.emb.embed_query(query)
        r = vs._collection.query(query_embeddings=[q], n_results=k, include=["metadatas", "embeddings"])
        embs = np.asarray(r["embeddings"][0])
        return [{"rank": i, "dom": dom_of(m), "cos": cosine(q, e)}
                for i, (m, e) in enumerate(zip(r["metadatas"][0], embs), 1)]

    # ---- Experiment A
    def run_separate(self, queries):
        rows = []
        for dom, vs in (("eco", self.eco_vs), ("soc", self.soc_vs)):
            for pid, q in enumerate(queries):
                hits = self.retrieve(vs, q)
                for K in K_LIST:
                    c = [h["cos"] for h in hits[:K]]
                    rows.append({"domain": dom, "prompt_id": pid, "K": K, "cos_mean": np.mean(c), "cos_rank1": hits[0]["cos"]})
        return pd.DataFrame(rows)

    @staticmethod
    def paired_diffs(df):
        out = {}
        for K, g in df.groupby("K"):
            e = g[g.domain == "eco"].set_index("prompt_id")["cos_mean"]
            s = g[g.domain == "soc"].set_index("prompt_id")["cos_mean"]
            out[K] = (e - s).dropna().to_numpy()
        return out

    def summarize_separate(self, df_task, df_ctrl):
        dt, dc = self.paired_diffs(df_task), self.paired_diffs(df_ctrl)
        rows = []
        for K in K_LIST:
            m, lo, hi = boot_ci(dt[K], self.rng)
            mc, loc, hic = boot_ci(dc[K], self.rng)
            corr = (self.rng.choice(dt[K], (N_BOOT, len(dt[K]))).mean(1)
                    - self.rng.choice(dc[K], (N_BOOT, len(dc[K]))).mean(1))
            rows.append({"K": K, "delta_task": m, "task_lo": lo, "task_hi": hi,
                         "delta_neutral_control": mc, "ctrl_lo": loc, "ctrl_hi": hic,
                         "delta_corrected": float(corr.mean()), "corr_lo": float(np.quantile(corr, .025)),
                         "corr_hi": float(np.quantile(corr, .975)), "P(corrected>0)": float((corr > 0).mean())})
        return pd.DataFrame(rows)

    # ---- Experiment B
    def run_mixed(self, queries):
        rows, rank_rows = [], []
        for pid, q in enumerate(queries):
            hits = self.retrieve(self.mixed_vs, q)
            dom = np.array([h["dom"] for h in hits])
            cos = np.array([h["cos"] for h in hits])
            for r, d in enumerate(dom, 1):
                rank_rows.append({"prompt_id": pid, "rank": r, "p_eco": d == "eco", "p_soc": d == "soc"})
            for K in K_LIST:
                d, c, w = dom[:K], cos[:K], w_dcg(K)
                e, s = d == "eco", d == "soc"
                rows.append({"prompt_id": pid, "K": K, "n_eco": int(e.sum()), "n_soc": int(s.sum()),
                             "eco_share": e.mean(), "eco_exposure": float((w * e).sum() / w.sum()),
                             "first_eco": (np.argmax(e) + 1) if e.any() else np.nan,
                             "first_soc": (np.argmax(s) + 1) if s.any() else np.nan,
                             "cos_eco": c[e].mean() if e.any() else np.nan,
                             "cos_soc": c[s].mean() if s.any() else np.nan})
        return pd.DataFrame(rows), pd.DataFrame(rank_rows)

    def summarize_mixed(self, df):
        rows = []
        for K, g in df.groupby("K"):
            r = {"K": K, "expected_share": self.expected_share}
            for c in ("eco_share", "eco_exposure", "first_eco", "first_soc"):
                r[c], r[c + "_lo"], r[c + "_hi"] = boot_ci(g[c], self.rng)
            r["eco_share_excess"] = r["eco_share"] - self.expected_share
            r["delta_cos_within_topK"] = float((g.cos_eco - g.cos_soc).dropna().mean())
            b = self.rng.choice(g.eco_share.to_numpy(), (N_BOOT, len(g))).mean(1)
            r["P(eco_share>expected)"] = float((b > self.expected_share).mean())
            rows.append(r)
        return pd.DataFrame(rows)

    # ---- full run
    def run(self, prompts, neutral):
        files = ("separate_summary.csv", "mixed_summary.csv", "rank_positions.csv")
        if not FORCE_RERUN and all(os.path.exists(f"{self.out_dir}/{f}") for f in files):
            print(f"results already exist in {self.out_dir} -> loading (set FORCE_RERUN=True to recompute)")
            return {"sep": pd.read_csv(f"{self.out_dir}/{files[0]}"),
                    "mix": pd.read_csv(f"{self.out_dir}/{files[1]}"),
                    "rank": pd.read_csv(f"{self.out_dir}/{files[2]}")}

        self.eco_vs, self.soc_vs = self.domain_store("eco"), self.domain_store("soc")
        self.mixed_vs = self.mixed_store({"eco": self.eco_vs, "soc": self.soc_vs})

        print("\n=== SANITY CHECKS ===")
        for nm, vs in (("eco", self.eco_vs), ("soc", self.soc_vs), ("mixed", self.mixed_vs)):
            stored_dim = len(vs._collection.get(limit=1, include=["embeddings"])["embeddings"][0])
            assert stored_dim == self.dim, f"{nm}: stored dim {stored_dim} != model dim {self.dim} (stale store from another model?)"
            print(f"{nm:6s} chunks={vs._collection.count():5d}  domains/files={self.domain_counts(vs)}  "
                  f"re-embed vs stored cosine={self.check_store(vs):.4f}   (expect ~1.0000)")
        cnt = self.domain_counts(self.mixed_vs)[0]
        assert set(cnt) <= {"eco", "soc"}, f"unknown domain tags in mixed store: {cnt}"
        self.expected_share = cnt.get("eco", 0) / max(1, cnt.get("eco", 0) + cnt.get("soc", 0))
        print(f"expected eco share of the mixed index = {self.expected_share:.3f}")

        print("\n=== EXPERIMENT A: separate corpora (task prompts + off-topic control) ===")
        df_task, df_ctrl = self.run_separate(prompts), self.run_separate(neutral)
        sum_sep = self.summarize_separate(df_task, df_ctrl)
        print(sum_sep.round(4).to_string(index=False))

        print("\n=== EXPERIMENT B: mixed corpus ===")
        df_mix, df_rank = self.run_mixed(prompts)
        sum_mix = self.summarize_mixed(df_mix)
        rank_sum = df_rank.groupby("rank")[["p_eco", "p_soc"]].mean().reset_index()
        print(sum_mix.round(4).to_string(index=False))
        print("\nP(chunk at rank r is ecological / social):")
        print(rank_sum.round(3).to_string(index=False))

        o = self.out_dir
        df_task.to_csv(f"{o}/separate_task_raw.csv", index=False)
        df_ctrl.to_csv(f"{o}/separate_control_raw.csv", index=False)
        sum_sep.to_csv(f"{o}/separate_summary.csv", index=False)
        df_mix.to_csv(f"{o}/mixed_raw.csv", index=False)
        sum_mix.to_csv(f"{o}/mixed_summary.csv", index=False)
        rank_sum.to_csv(f"{o}/rank_positions.csv", index=False)
        self.plot(sum_sep, sum_mix, rank_sum)
        print("\nSaved to:", o)
        return {"sep": sum_sep, "mix": sum_mix, "rank": rank_sum}

    def plot(self, sum_sep, sum_mix, rank_sum):
        fig, ax = plt.subplots(1, 3, figsize=(16, 4))
        ax[0].errorbar(sum_sep.K, sum_sep.delta_task, yerr=[sum_sep.delta_task - sum_sep.task_lo, sum_sep.task_hi - sum_sep.delta_task],
                       marker="o", capsize=3, label="task prompts")
        ax[0].errorbar(sum_sep.K, sum_sep.delta_neutral_control, yerr=[sum_sep.delta_neutral_control - sum_sep.ctrl_lo,
                       sum_sep.ctrl_hi - sum_sep.delta_neutral_control], marker="s", capsize=3, label="off-topic control")
        ax[0].plot(sum_sep.K, sum_sep.delta_corrected, "k--", marker="^", label="corrected (task - control)")
        ax[0].axhline(0, color="r", lw=.8); ax[0].axvline(BASE_K, color="gray", ls=":")
        ax[0].set_xlabel("K"); ax[0].set_ylabel("Δ cosine (eco − soc)"); ax[0].set_title("Separate corpora"); ax[0].legend(fontsize=7)

        ax[1].plot(sum_mix.K, sum_mix.eco_share, marker="o", label="eco share")
        ax[1].fill_between(sum_mix.K, sum_mix.eco_share_lo, sum_mix.eco_share_hi, alpha=.2)
        ax[1].plot(sum_mix.K, sum_mix.eco_exposure, marker="s", label="eco exposure (rank-weighted)")
        ax[1].axhline(self.expected_share, color="k", ls="--", label="index share (neutral)")
        ax[1].axvline(BASE_K, color="gray", ls=":"); ax[1].set_xlabel("K"); ax[1].set_title("Mixed corpus: who wins top-K")
        ax[1].legend(fontsize=7)

        ax[2].bar(rank_sum["rank"], rank_sum.p_eco, label="eco")
        ax[2].bar(rank_sum["rank"], rank_sum.p_soc, bottom=rank_sum.p_eco, label="social")
        ax[2].set_xlabel("rank position"); ax[2].set_ylabel("share of prompts"); ax[2].set_title("Domain at each rank"); ax[2].legend(fontsize=7)
        fig.suptitle(self.label)
        plt.tight_layout(); plt.savefig(f"{self.out_dir}/rank_k_figures.png", dpi=200); plt.show()


# ----------------------------------------------------------------------------- cross-run comparison
def compare(results):
    rows = []
    for lab, r in results.items():
        m, s, rk = r["mix"], r["sep"], r["rank"]
        exp = float(m.expected_share.iloc[0])
        at = lambda df, col, K: float(df.loc[df.K == K, col].iat[0])
        below = m[m.eco_share < exp]
        sig_K = 0                                               # largest K up to which corrected CI stays > 0
        for K in sorted(s.K):
            if float(s.loc[s.K == K, "corr_lo"].iat[0]) > 0:
                sig_K = K
            else:
                break
        rows.append({
            "run": lab,
            "ctrl_offset_mean": float(s.delta_neutral_control.mean()),
            "dcorr@1": at(s, "delta_corrected", 1), "dcorr@4": at(s, "delta_corrected", 4), "dcorr@10": at(s, "delta_corrected", 10),
            "dcorr_CI>0_up_to_K": sig_K,
            "eco_share@1": at(m, "eco_share", 1), "eco_share@4": at(m, "eco_share", 4),
            "eco_share@8": at(m, "eco_share", 8), "eco_share@20": at(m, "eco_share", 20),
            "eco_exposure@4": at(m, "eco_exposure", 4),
            "expected_share": exp,
            "crossover_K(eco_share<expected)": int(below.K.min()) if len(below) else None,
            "p_eco_rank1": float(rk.loc[rk["rank"] == 1, "p_eco"].iat[0]),
            "p_eco_ranks1-4": float(rk.loc[rk["rank"] <= 4, "p_eco"].mean()),
            "p_eco_ranks5-20": float(rk.loc[rk["rank"] > 4, "p_eco"].mean()),
        })
    cmp_df = pd.DataFrame(rows)
    print("\n" + "=" * 90 + "\nCROSS-RUN COMPARISON (compare patterns, NOT raw cosine magnitudes across models)\n" + "=" * 90)
    print(cmp_df.round(3).T.to_string(header=False))
    cmp_df.to_csv(f"{BASE}/rank_k_comparison.csv", index=False)

    fig, ax = plt.subplots(1, 3, figsize=(16, 4))
    for lab, r in results.items():
        ax[0].plot(r["mix"].K, r["mix"].eco_share, marker="o", label=lab)
        ax[1].plot(r["sep"].K, r["sep"].delta_corrected, marker="^", label=lab)
        ax[2].plot(r["rank"]["rank"], r["rank"].p_eco, marker=".", label=lab)
    ax[0].axhline(float(next(iter(results.values()))["mix"].expected_share.iloc[0]), color="k", ls="--")
    ax[0].set_title("Mixed: eco share vs K"); ax[0].set_xlabel("K")
    ax[1].axhline(0, color="r", lw=.8); ax[1].set_title("Separate: control-corrected Δ vs K"); ax[1].set_xlabel("K")
    ax[2].axhline(0.5, color="k", ls="--", lw=.8); ax[2].set_title("P(chunk at rank r is eco)"); ax[2].set_xlabel("rank")
    ax[0].legend(fontsize=7)
    plt.tight_layout(); plt.savefig(f"{BASE}/rank_k_comparison.png", dpi=200); plt.show()
    print("Saved comparison to:", BASE)
    return cmp_df


# ----------------------------------------------------------------------------- main
RESULTS = {}
for model, use_prefix in RUNS:
    pipe = Pipeline(model, use_prefix)
    RESULTS[pipe.label] = pipe.run(PROMPTS, NEUTRAL)
CMP = compare(RESULTS) if len(RESULTS) > 1 else None
````
